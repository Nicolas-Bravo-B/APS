# Análisis del Circuito a)

## 1. Función de Transferencia $H(s)$

A partir del divisor de tensión del circuito (donde la impedancia del capacitor es $Z_C = \frac{1}{sC}$ en paralelo con la salida, y en serie se encuentran la inductancia $L$ y la resistencia $R$):

$$H(s) = \frac{\frac{1}{sC}}{\frac{1}{sC} + sL + R}$$

Trabajando algebraicamente la expresión multiplicando numerador y denominador por $sC$:

$$H(s) = \frac{\frac{1}{sC}}{\frac{1}{sC}\left(s^2LC + sRC + 1\right)} = \frac{\frac{1}{LC}}{s^2 + s\frac{R}{L} + \frac{1}{LC}}$$

Reemplazando los valores dados ($\omega_0 = 1$, $R = 1\,\Omega$ y $Q = \frac{\sqrt{2}}{2}$), sabiendo que $\frac{\omega_0}{Q} = \frac{R}{L}$ y $\omega_0^2 = \frac{1}{LC}$ (lo que resulta en $L = \frac{\sqrt{2}}{2}$ y $C = \sqrt{2}$):

$$H(s) = \frac{1}{s^2 + \sqrt{2}s + 1}$$

---

## 2. Respuesta en Frecuencia $H(j\omega)$

Reemplazando $s = j\omega$:

$$H(j\omega) = \frac{1}{-\omega^2 + 1 + j\omega\sqrt{2}}$$

### Módulo de la Función de Transferencia
$$|H(j\omega)| = \frac{1}{\sqrt{(1-\omega^2)^2 + (\omega\sqrt{2})^2}} = \frac{1}{\sqrt{1 - 2\omega^2 + \omega^4 + 2\omega^2}} = \frac{1}{\sqrt{\omega^4 + 1}}$$

Evaluando en puntos notables:
* Para $\omega = 0$: $|H|_{0} = 1$
* Para $\omega = \omega_0 = 1$: $|H|_{j\omega_1} = \frac{1}{\sqrt{2}} \approx 0.707$
* Para $\omega \to \infty$: $|H|_{\omega \to \infty} = 0$

### Fase de la Función de Transferencia
$$\arg(H(j\omega)) = \Delta H = 0 - \operatorname{tg}^{-1}\left(\frac{\omega\sqrt{2}}{1-\omega^2}\right)$$

Evaluando en puntos notables:
* Para $\omega = 0$: $\Delta H_0 = -\operatorname{tg}^{-1}(0) = 0$
* Para $\omega = 1$: $\Delta H_1 = -\operatorname{tg}^{-1}\left(\frac{\sqrt{2}}{0}\right) = -\frac{\pi}{2}$
* Para $\omega \to \infty$: $\Delta H_{\omega \to \infty} = -\operatorname{tg}^{-1}\left(\frac{\omega\sqrt{2}}{1-\omega^2}\right) = -\operatorname{tg}^{-1}\left(\frac{\sqrt{2}/\omega}{\frac{1}{\omega^2} - 1}\right) \to -\pi$ (tendiendo por cuadrante a $-\pi$).

---

## 3. Ancho de Banda (Bw) y Tipo de Filtro

El ancho de banda se define para los valores donde el módulo cae a $\frac{1}{\sqrt{2}}$ de su valor máximo:

$$|H(j\omega)| = \frac{1}{\sqrt{2}} \implies |H(j\omega)|^2 = \frac{1}{2}$$

$$\frac{1}{\omega^4 + 1} = \frac{1}{2} \implies \omega^4 + 1 = 2 \implies \omega^4 - 1 = 0$$

Resolviendo para la frecuencia de corte $\omega_c$:
$$\omega_c = 1$$

Por lo tanto, el ancho de banda del filtro es:
$$\text{Bw} = [0, 1]$$

**Conclusión:** Al permitir el paso de las frecuencias bajas desde $\omega = 0$ hasta $\omega = 1$ y atenur las altas frecuencias, se trata de un **Filtro Pasabajos**.

---

3. **Resistencia ($R$):**

# Análisis del Circuito b)

## 1. Función de Transferencia $H(s)$

A partir del circuito b, planteamos las admitancias e impedancias de los componentes:
* $Z_L = sL$, $Z_C = \frac{1}{sC}$, $Z_R = R$
* $Y_1 = \frac{1}{R} + sC$
* $Y_2 = \frac{1}{sL}$

Aplicando divisor de tensión:
$$V_2 = V_1 \cdot \frac{Y_1}{Y_1 + Y_2} \implies H(s) = \frac{V_2}{V_1} = \frac{Y_1}{Y_1 + Y_2}$$

Sustituyendo los valores:
$$H(s) = \frac{\frac{1}{R} + sC}{\frac{1}{R} + sC + \frac{1}{sL}} = \frac{\frac{1}{R} + sC}{\frac{1}{sL} \cdot \left(s^2LC + s\frac{L}{R} + 1\right)}$$

Operando algebraicamente para llevarlo a la forma canónica:
$$H(s) = \frac{s^2LC + s\frac{L}{R}}{LC \left(s^2 + s\frac{1}{RC} + \frac{1}{LC}\right)} = \frac{s^2 + \frac{s}{RC}}{s^2 + \frac{s}{RC} + \frac{1}{LC}}$$

Reemplazando los parámetros ($\omega_0 = 1$, $R = 1\,\Omega$, $Q = \frac{\sqrt{2}}{2}$), se obtienen los valores normalizados ($L = \sqrt{2}$, $C = \frac{\sqrt{2}}{2}$):
$$H(s) = \frac{s^2 + \sqrt{2}s}{s^2 + \sqrt{2}s + 1}$$

---

## 2. Respuesta en Frecuencia $H(j\omega)$

Reemplazando $s = j\omega$:
$$H(j\omega) = \frac{-\omega^2 + j\omega\sqrt{2}}{-\omega^2 + j\omega\sqrt{2} + 1}$$

### Módulo y Fase
$$|H(j\omega)| = \frac{\sqrt{\omega^4 + \omega^2}}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{\sqrt{\omega^4 + \omega^2}}{\sqrt{\omega^4 + 1}}$$

* **Evaluación en puntos notables:**
  * $|H(0)| = 0$
  * $|H(1)| = \frac{\sqrt{2}}{\sqrt{2}} = 1$
  * $\lim_{\omega \to \infty} |H(j\omega)| = 1$

* **Fase ($\Delta H$):**
  $$\Delta H = \tan^{-1}\left(\frac{\sqrt{2}}{\omega}\right) - \tan^{-1}\left(\frac{\omega\sqrt{2}}{1-\omega^2}\right)$$
  * $\Delta H(0) = \frac{\pi}{2} = 90^\circ$
  * $\Delta H(1) = 35.26^\circ$
  * $\lim_{\omega \to \infty} \Delta H = -\frac{\pi}{2} = -90^\circ$

---

## 3. Ancho de Banda y Tipo de Filtro (Circuito b)

Para hallar el ancho de banda, igualamos $|H(\omega_c)| = \frac{1}{\sqrt{2}}$:
$$\frac{\omega^4 + \omega^2}{\omega^4 + 1} = \frac{1}{2} \implies 2\omega^4 + 2\omega^2 = \omega^4 + 1 \implies \omega^4 + 2\omega^2 - 1 = 0$$

Haciendo el cambio de variable $\sigma = \omega^2$:
$$\sigma^2 + 2\sigma - 1 = 0 \implies \sigma_1 = 0.2247 \quad (\sigma_2 = -2.2247 \text{ descartada})$$

Calculando la frecuencia de corte:
$$\omega_c = \sqrt{0.2247} \approx 0.426$$

**Conclusión:** Se trata de un **Filtro Pasalto** cuyo ancho de banda se extiende desde $\omega_c$ hasta infinito ($\text{Bw} = [0.426, \infty)$).

---

# Análisis del Circuito c)

## 1. Función de Transferencia $H(s)$

A partir del circuito c, las admitancias e impedancias son:
* $Z_R = R$, $Z_L = sL$, $Z_C = \frac{1}{sC}$
* $Y_1 = \frac{1}{R}$
* $Y_2 = \frac{1}{sL} + sC$

Aplicando divisor de tensión:
$$H(s) = \frac{Y_1}{Y_1 + Y_2} = \frac{\frac{1}{R}}{\frac{1}{R} + \frac{1}{sL} + sC}$$

Operando algebraicamente:
$$H(s) = \frac{\frac{1}{sL}}{\frac{1}{sL \cdot R}\left(s^2LC + s\frac{L}{R} + 1\right)} = \frac{\frac{s}{RC}}{s^2 + \frac{s}{RC} + \frac{1}{LC}}$$

Sustituyendo los valores normalizados ($L = \sqrt{2}$, $C = \frac{\sqrt{2}}{2}$, $R = 1$):
$$H(s) = \frac{\sqrt{2}s}{s^2 + \sqrt{2}s + 1}$$

---

## 2. Respuesta en Frecuencia $H(j\omega)$

Reemplazando $s = j\omega$:
$$H(j\omega) = \frac{j\omega\sqrt{2}}{-\omega^2 + j\omega\sqrt{2} + 1}$$

### Módulo y Fase
$$|H(j\omega)| = \frac{\omega\sqrt{2}}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{\omega\sqrt{2}}{\sqrt{\omega^4 + 1}}$$

* **Evaluación en puntos notables:**
  * $|H(0)| = 0$
  * $|H(1)| = \frac{\sqrt{2}}{\sqrt{2}} = 1$
  * $\lim_{\omega \to \infty} |H(j\omega)| = 0$

* **Fase ($\Delta H$):**
  $$\Delta H = \frac{\pi}{2} - \tan^{-1}\left(\frac{\omega\sqrt{2}}{1-\omega^2}\right)$$
  * $\Delta H(0) = \frac{\pi}{2} = 90^\circ$
  * $\Delta H(1) = 0^\circ$
  * $\lim_{\omega \to \infty} \Delta H = -\frac{\pi}{2} = -90^\circ$

---

## 3. Ancho de Banda y Tipo de Filtro (Circuito c)

Igualando el módulo a $\frac{1}{\sqrt{2}}$ para hallar las frecuencias de corte:
$$\frac{\omega\sqrt{2}}{\sqrt{\omega^4 + 1}} = \frac{1}{\sqrt{2}} \implies 2\omega^2 = \omega^4 + 1 \implies \omega^4 - 2\omega^2 + 1 = 0$$

Haciendo $\sigma = \omega^2$:
$$\sigma^2 - 2\sigma + 1 = 0 \implies \sigma_1 = 3.73 \quad (\text{o bien } \sigma_2 = 0.2679)$$

Calculando las frecuencias de corte $\omega_{c1}$ y $\omega_{c2}$:
* $\omega_{c1} = \sqrt{3.73} = 1.93$
* $\omega_{c2} = \sqrt{0.2679} = 0.51$

**Conclusión:** Se trata de un **Filtro Pasabanda** cuyo ancho de banda va desde $\omega_{c2} = 0.51$ hasta $\omega_{c1} = 1.93$ ($\text{Bw} = [0.51, 1.93]$).

---

# Análisis del Circuito Bonus

## 1. Función de Transferencia $H(s)$

A partir del circuito bonus, planteamos las admitancias de los bloques:
* $Y_1 = \frac{b}{sL} + sCa$
* $Y_2 = \frac{1}{R} + \frac{1-b}{sL} + sC(1-a)$

Aplicando divisor de tensión:
$$H(s) = \frac{Y_1}{Y_1 + Y_2} = \frac{\frac{b}{sL} + sCa}{\frac{b}{sL} + sCa + \frac{1}{R} + \frac{1-b}{sL} + sC(1-a)}$$

Operando algebraicamente para llevarlo a la forma canónica:
$$H(s) = \frac{\frac{b}{sL} + sCa}{\frac{1}{sL}\left(s^2LC + s\frac{L}{R} + 1\right)} = \frac{a \left(s^2 + \frac{b}{LCa}\right)}{s^2 + s\frac{L}{R} + \frac{1}{LC}}$$

Sustituyendo los parámetros normalizados ($L = \sqrt{2}$, $C = \frac{\sqrt{2}}{2}$, $R = 1$), se obtiene la expresión general:
$$H(s) = \frac{a \left(s^2 + \frac{b}{a}\right)}{s^2 + \sqrt{2}s + 1}$$

---

## 2. Respuesta en Frecuencia $H(j\omega)$

Reemplazando $s = j\omega$:
$$H(j\omega) = \frac{a \left(-\omega^2 + \frac{b}{a}\right)}{-\omega^2 + j\omega\sqrt{2} + 1}$$

### Módulo y Fase
$$|H(j\omega)| = \frac{\left|a \left(-\omega^2 + \frac{b}{a}\right)\right|}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{\left|a \left(-\omega^2 + \frac{b}{a}\right)\right|}{\sqrt{\omega^4 + 1}}$$

* **Evaluación en puntos notables:**
  * Para $\omega = 0$: $|H(0)| = b$
  * Para $\omega = 1$: $|H(1)| = \frac{\left|-a + b\right|}{\sqrt{2}}$
  * Para $\omega \to \infty$: $\lim_{\omega \to \infty} |H(j\omega)| = a$

---

## 3. Ancho de Banda y Tipo de Filtro (Circuito Bonus)

El comportamiento y el ancho de banda del circuito dependen directamente de la relación entre los parámetros $a$ y $b$:

* **Si $a = b$:** Se trata de un **Filtro Notch (Rechaza Banda)**. Atenúa fuertemente en la frecuencia de sintonía $\omega_0 = \sqrt{\frac{b}{a}} = 1$, presentando un ancho de banda de rechazo angosto en torno a dicha frecuencia y permitiendo el paso de las bandas inferiores y superiores.
* **Si $a \neq b$:** Se trata de un **Filtro de paso con escalón (Modified Notch)**, el cual presenta una muesca de atenuación pero con distintos niveles de ganancia asintótica entre bajas frecuencias ($|H(0)| = b$) y altas frecuencias ($\lim_{\omega \to \infty} |H| = a$).
* **Si $a = 0$ y $b = 1$:** Se reduce a la función de transferencia del circuito base, tratándose de un **Filtro Pasa-bajos**. Atenúa las frecuencias superiores a $\omega_0 = 1$ y permite el paso de las frecuencias bajas desde continua hasta la frecuencia de corte.
* **Si $a = 1$ y $b = 0$:** Se reduce a la función de transferencia del circuito de alta frecuencia, tratándose de un **Filtro Pasa-altos**. Atenúa las frecuencias inferiores a $\omega_0 = 1$ y permite el paso de las componentes de alta frecuencia hacia la salida.
